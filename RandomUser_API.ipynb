```python
from randomuser import RandomUser
import pandas as pd
```

```python
# created an empty rnd variable

rnd = RandomUser()
```

```python
# generated 10 random users in rnd variable

list_of_10 = rnd.generate_users(10)
```

```python
# list down names and emails via loop (iterative process)

for user in list_of_10:
    print (user.get_full_name()," ",user.get_email())
```

```python
# list down picture's URL via loop (iterative process)

for user in list_of_10:
    print (user.get_picture())
```

```python
# created a function that generates updated 10 keys and their values every time whenever calls

def get_users():
    users =[]
     
    for user in RandomUser.generate_users(10):
        users.append({"Name":user.get_full_name(),"Gender":user.get_gender(),"City":user.get_city(),"State":user.get_state(),"Email":user.get_email(), "DOB":user.get_dob(),"Picture":user.get_picture()})
      
    return pd.DataFrame(users)     
```
